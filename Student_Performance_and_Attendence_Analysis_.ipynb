```python
# Student Performance and Attendence Analysis
import pandas as pd

df = pd.read_csv("mark_sheet_clean_file1 (1).csv")

print("STUDENT PERFORMANCE AND ATTENDANCE SYSTEM")

print("Total Students:", len(df))

n = int(input("Enter Student Number (1-1005): "))

if n >= 1 and n <= len(df):

    student = df.iloc[n - 1]

    math = student["Math_Score"]
    english = student["English_Score"]
    science = student["Science_Score"]
    attendance = student["Attendance"]

    total = math + english + science
    average = total / 3

    print("\n--- Student Details ---")
    print("Math Marks:", round(math, 2))
    print("English Marks:", round(english, 2))
    print("Science Marks:", round(science, 2))

    print("Total Marks:", round(total, 2))
    print("Average Marks:", round(average, 2))
    print("Attendance:", round(attendance, 2), "%")

    if attendance >= 75:
        print("Attendance Status: Good")
    else:
        print("Attendance Status: Low")

else:
    print("Invalid Student Number")
```
